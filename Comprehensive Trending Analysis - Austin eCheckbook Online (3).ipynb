#### Data Preprocessing and Feature Engineering!

```python
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)
pd.set_option('display.max_colwidth', None)

# Convert AMOUNT to numeric (cleaning commas if present)
df['AMOUNT'] = pd.to_numeric(df['AMOUNT'].astype(str).str.replace(',', ''), errors='coerce')
# Convert Date column to datetime & extract Year / Month
df['CHK_EFT_ISS_DT'] = pd.to_datetime(df['CHK_EFT_ISS_DT'], errors='coerce')
df['TRANS_YEAR'] = df['CHK_EFT_ISS_DT'].dt.year
df['TRANS_MONTH'] = df['CHK_EFT_ISS_DT'].dt.month
# 1. Total Spending by Department
dept_spending = df.groupby('DEPT_NM')['AMOUNT'].sum().sort_values(ascending=False).reset_index()
# 2. All Payees / Vendors (Sorted by highest spend)
# Note: Use 'VEND_CUST_NM' instead of 'VEND_CUST_CD' if you want Vendor Names instead of IDs
top_vendors = df.groupby('VEND_CUST_CD')['AMOUNT'].sum().sort_values(ascending=False).reset_index()
# 3. Monthly Spend Trends over Time
monthly_trends = df.groupby(['TRANS_YEAR', 'TRANS_MONTH'])['AMOUNT'].sum().reset_index()
# --- Display Results ---
print("--- Department Spending ---")
display(dept_spending)
print("\n--- All Vendors / Payees ---")
display(top_vendors)
print("\n--- Monthly Trends ---")
display(monthly_trends)
```

```python
df['AMOUNT'] = pd.to_numeric(df['AMOUNT'].astype(str).str.replace(',', ''), errors='coerce')
df=df.dropna(subset=['AMOUNT'])
#Resetting index after data cleaning.
df=df.reset_index(drop=True)
df['CHK_EFT_ISS_DT'] = pd.to_datetime(df['CHK_EFT_ISS_DT'], errors='coerce')
df['TRANS_YEAR'] = df['CHK_EFT_ISS_DT'].dt.year
df['TRANS_MONTH'] = df['CHK_EFT_ISS_DT'].dt.month
df['DAY_OF_WEEK'] = df['CHK_EFT_ISS_DT'].dt.day_name()
def get_amount_category(amt):
    if pd.isna(amt):
        return 'Unknown'
    elif amt < 1000:
        return 'Small'
    elif amt <= 50000:
        return 'Medium'
    else:
        return 'Large'

df['AMOUNT_CATEGORY'] = df['AMOUNT'].apply(get_amount_category)
print(df[['AMOUNT', 'TRANS_YEAR', 'TRANS_MONTH', 'DAY_OF_WEEK', 'AMOUNT_CATEGORY']].head())
```

### Find Outliers in Dataset!

```python
# Ensure visualization style is clean
sns.set_theme(style="whitegrid")
## To remove negative values
df=df[df['AMOUNT']>=0]

# ==========================================
# 1. MATHEMATICAL IQR CALCULATION
# ==========================================

# Calculate Quartiles and IQR
Q1 = df['AMOUNT'].quantile(0.25)
Q3 = df['AMOUNT'].quantile(0.75)
IQR = Q3 - Q1

# Define Upper and Lower Limits (Fences)
lower_fence = max(0, Q1 - 1.5 * IQR)
upper_fence = Q3 + 1.5 * IQR

print(f"--- IQR Summary Statistics ---")
print(f"25th Percentile (Q1): ${Q1:,.2f}")
print(f"75th Percentile (Q3): ${Q3:,.2f}")
print(f"IQR Value:            ${IQR:,.2f}")
print(f"Lower Bound Fence:    ${lower_fence:,.2f}")
print(f"Upper Bound Fence:    ${upper_fence:,.2f}\n")

# Identify anomalies vs normal records
outliers_df = df[(df['AMOUNT'] < lower_fence) | (df['AMOUNT'] > upper_fence)]
cleaned_df = df[(df['AMOUNT'] >= lower_fence) & (df['AMOUNT'] <= upper_fence)]


print(f"Original Row Count: {len(df)}")
print(f"Outlier Rows Found:  {len(outliers_df)} ({len(outliers_df)/len(df)*100:.2f}%)")
print(f"Cleaned Row Count:   {len(cleaned_df)}\n")

# ==========================================
# 2. PLOTTING VISUAL COMPARISONS (FIGURES)
# ==========================================

# Create a side-by-side subplot figure
fig, axes = plt.subplots(1, 2, figsize=(15, 6))

# Figure A: Before Outlier Removal
sns.boxplot(data=df, y='AMOUNT', ax=axes[0], color='#d9534f')
axes[0].set_title('Figure 1: Expenditure Distribution (With Outliers)', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Transaction Amount ($)')
axes[0].set_xlabel('Raw Data')
# Figure B: After Outlier Removal
sns.boxplot(data=cleaned_df, y='AMOUNT', ax=axes[1], color='#5c8585')
axes[1].set_title('Figure 2: Expenditure Distribution (Cleaned via IQR)', fontsize=13, fontweight='bold')
axes[1].set_ylabel('Transaction Amount ($)')
axes[1].set_xlabel('Filtered Data (Within Fences)')
axes[1].set_ylim(bottom=0)
# Optimize geometry and render
plt.subplots_adjust(left=0.08,right=0.95,wspace=0.3)
plt.tight_layout()
plt.show()
```
